```python
import os
os.environ["TORCHDYNAMO_DISABLE"] = "1"
os.environ["TORCH_COMPILE_DISABLE"] = "1"
os.environ["PYTORCH_JIT"] = "0"
os.environ["TORCHINDUCTOR_COMPILE_THREADS"] = "1"
os.environ["TORCHINDUCTOR_CPP_WRAPPER"] = "0"

import torch
torch._dynamo.config.suppress_errors = True
torch._dynamo.reset()

from docling.document_converter import DocumentConverter
from pathlib import Path
```

```python
RAW_PDF_DIR = Path("data/raw_pdfs")
converter = DocumentConverter()
result = converter.convert(str(RAW_PDF_DIR / "myrelief_claims_sop.pdf"))
print("Conversion done")
```

```python
markdown_output = result.document.export_to_markdown()
print(f"Length: {len(markdown_output)} characters")
print("---")
print(markdown_output[:2000])
```

```python
raw_pdf_dir = Path("data/raw_pdfs")
pdf_files = list(raw_pdf_dir.glob("*.pdf"))
print(len(pdf_files))
for f in pdf_files:
    print(" -", f.name)
```

```python
all_results = {}
for pdf_path in sorted(RAW_PDF_DIR.glob("*.pdf")):
    print(f"Converting {pdf_path.name}...")
    result = converter.convert(str(pdf_path))
    md = result.document.export_to_markdown()
    all_results[pdf_path.name] = md
    print(f"  -> {len(md)} characters\n")

print("All done:", list(all_results.keys()))
```

```python
for name, md in all_results.items():
    print(f"=== {name} ===")
    print(md[:500])
    print("...\n")
```
